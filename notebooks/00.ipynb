```python
products = [
    {"name": "키보드", "price": 45000, "stock": 12, "category": "전자기기"},
    {"name": "마우스", "price": 28000, "stock": 0, "category": "전자기기"},
    {"name": "모니터", "price": 320000, "stock": 4, "category": "전자기기"},
    {"name": "책상", "price": 180000, "stock": 0, "category": "가구"},
    {"name": "의자", "price": 150000, "stock": 7, "category": "가구"},
    {"name": "키보드", "price": 45000, "stock": 1, "category": "전자기기"},
]
```

```python
for_sales = []

found = False

for product in products:
    if product["stock"] >= 1:
        found = False
        for sale in for_sales:
            if sale["name"] == product["name"]:
                sale["stock"] += product["stock"]
                found = True
                break
        if not found:
            for_sales.append(product.copy())


for_sales
```
