## 실습 3. 분석에 필요한 컬럼만 선택하기

원본 데이터의 모든 컬럼이 이후 분석에 필요한 것은 아닙니다.

원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.

- 변경 후 컬럼: 순위, 판매상품ID, 상품명, 판매가, 저자, 출판사, 발행일, 분야

## 실습 5. 결측치 확인하기

결측치는 값이 비어 있는 데이터를 의미합니다.

데이터 분석에서는 결측치를 무조건 삭제하지 않습니다.

먼저 어디에 얼마나 있는지 확인하고, 분석 목적에 맞게 처리 방법을 결정합니다.

```python
df_books.isna().sum()  # 컬럼별로 결측치(비어있는 값) 개수를 세어 보여줍니다
```

```python
missing = df_books.isna().sum()  # 컬럼별 결측치 개수를 missing에 저장합니다

missing[missing > 0]  # 그중에서 개수가 0보다 큰(결측치가 있는) 컬럼만 골라 보여줍니다
```

### 결측치 확인 결과

- 결측치가 있는 컬럼은 총 4개입니다: 저자(3), 출판사(3), 발행일(3), 분야(62)

## 실습 6. 판매가를 숫자로 바꾸기

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.

```python
# 1단계: 쉼표(,)를 빈 문자열로 바꿔서 제거합니다
df_books['판매가'] = df_books['판매가'].str.replace(',', '')

# 2단계: 문자열을 정수형(int)으로 변환합니다
df_books['판매가'] = df_books['판매가'].astype(int)

# 3단계: 변환 후 자료형을 확인합니다
print(df_books['판매가'].dtype)

df_books.head()  # 실제 값도 눈으로 확인합니다
```

### 판매가 숫자형 변환 결과

- `판매가` 컬럼의 쉼표를 제거한 뒤 정수형(int64)으로 변환했습니다.
- 변환 후 dtype이 `int64`로 확인되어, 이제 판매가로 계산(합계, 평균, 비교 등)이 가능합니다.
- `head()`로 확인한 값들도 `16200`, `17820`처럼 쉼표 없이 정상적으로 숫자로 표시됩니다.

## 실습 7. 발행일을 날짜형으로 바꾸기

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

```python
# 1단계: 결측치를 허용하는 정수형(Int64)으로 바꿔서 소수점을 없앱니다
df_books['발행일'] = df_books['발행일'].astype('Int64')

# 2단계: 문자열로 바꾼 뒤, YYYYMMDD 형식에 맞춰 날짜로 변환합니다
# errors='coerce'를 넣으면 형식이 안 맞거나 결측치인 값은 에러 대신 NaT(빈 날짜)로 처리됩니다
df_books['발행일'] = pd.to_datetime(df_books['발행일'].astype(str), format='%Y%m%d', errors='coerce')

# 3단계: 변환 후 자료형과 값을 확인합니다
print(df_books['발행일'].dtype)
df_books['발행일'].head()
```

### 발행일 날짜형 변환 결과

- `발행일` 컬럼을 `datetime64[us]` 자료형으로 성공적으로 변환했습니다.
- 결측치였던 3개 값은 에러 없이 `NaT`(빈 날짜)로 처리되었습니다.
- 이제 `발행일`을 기준으로 연도/월별 집계나 기간 필터링 등의 날짜 연산이 가능합니다.

## 실습 8. 문자열 앞뒤 공백 정리하기

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.

```python
cols_to_strip = ['상품명', '저자', '출판사', '분야']  # 공백을 제거할 컬럼 이름들을 리스트로 정리합니다

for col in cols_to_strip:  # 리스트에 담긴 컬럼 이름을 하나씩 꺼내면서 반복합니다
    df_books[col] = df_books[col].str.strip()  # 해당 컬럼의 앞뒤 공백을 제거합니다

df_books[cols_to_strip].head()  # 처리 후 4개 컬럼의 앞 5행을 확인합니다
```

### 문자열 공백 제거 결과

- `상품명`, `저자`, `출판사`, `분야` 4개 컬럼에 반복문으로 `str.strip()`을 적용해 앞뒤 공백을 제거했습니다.
- `head()`로 확인한 결과 값들이 정상적으로 표시되며, 처리가 잘 적용되었습니다.
- 이제 이 컬럼들로 그룹화나 값 비교를 할 때 공백 때문에 같은 값이 다르게 인식되는 문제를 방지할 수 있습니다.

## 실습 9. 중복 데이터 확인하기

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.

하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

우선 상품을 구분하는 판매상품ID를 확인합니다.

```python
# 1. 전체 행이 완전히 같은 중복 행 개수
print(df_books.duplicated().sum())

# 2. 판매상품ID가 중복된 행 개수
print(df_books.duplicated(subset='판매상품ID').sum())

# 3. 상품명이 중복된 행 개수
print(df_books.duplicated(subset='상품명').sum())
```

### 중복 데이터 확인 결과

- 전체 행이 완전히 같은 중복: 0건
- 판매상품ID 기준 중복: 0건 → 상품을 구분하는 고유 식별자로 신뢰할 수 있습니다.
- 상품명 기준 중복: 37건 → 같은 제목이 여러 번 등장하지만, 다른 판본·개정판일 수 있으므로 바로 삭제하지 않습니다.
